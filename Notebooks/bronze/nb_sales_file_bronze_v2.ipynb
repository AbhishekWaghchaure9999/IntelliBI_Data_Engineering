```python
dfMonthlySales = spark.read.format("csv").option("header", "true").load("/Volumes/intellibicatalogo/intellibi_schema/intellibi_data_engineering/")
display(dfMonthlySales);
```

```python
dfMonthlySalesClean = dfMonthlySales.selectExpr(
    'trim(Customer_ID) as Customer_ID',
    'trim(Customer_Name) as Customer_Name',
    'trim(Gender) as Gender',
    'cast(trim(Age) as integer) as Age',
    'trim(City) as City',
    'trim(custpurPrice) as custpurPrice',
    'trim(purchase_date) as purchase_date',
    'current_date() as ingest_ts' 
)

dfMonthlySalesClean.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("intellibicatalogo.intellibi_bronze.dfMonthlySales")
```
