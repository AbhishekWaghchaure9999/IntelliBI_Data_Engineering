```python
dfMonthlySales = spark.read.format("csv").option("header", "true").load("/Volumes/intellibicatalogo/intellibi_schema/intellibi_data_engineering/CustDetails.csv")
display(dfMonthlySales)
```

```python
from pyspark.sql.functions import col

dfMonthlySales_typed = dfMonthlySales \
    .withColumn("Customer_ID", col("Customer_ID").cast("integer")) \
    .withColumn("Age", col("Age").cast("integer")) \
    .withColumn("custpurPrice", col("custpurPrice").cast("double")) \
    .withColumn("purchase_date", col("purchase_date").cast("date"))

dfMonthlySales.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("intellibicatalogo.intellibi_schema.dfMonthlySales")
```
