```python
%sql
-- inserting data from bronze to silver

INSERT INTO intellibicatalogo.intellibi_silver.cleanedMonthlySales
SELECT DISTINCT
    TRIM(Customer_ID) AS Customer_ID,
    UPPER(TRIM(Customer_Name)) AS Customer_Name,
    TRIM(Gender) AS Gender,  
    CAST(TRIM(Age) AS INTEGER) AS Age,
    UPPER(TRIM(City)) AS City,
    CAST(TRIM(custpurPrice) AS INTEGER) AS custpurPrice,
    TO_DATE(TRIM(purchase_date), 'yyyy-MM-dd') AS purchase_date,
    CURRENT_TIMESTAMP() AS ingest_ts
FROM intellibicatalogo.intellibi_bronze.rawMonthlySales
WHERE ingest_ts > (
                    SELECT COALESCE(MAX(ingest_ts), TO_DATE(1900-01-01), 'YYYY-MM-DD') 
                    FROM intellibicatalogo.intellibi_silver.cleanedMonthlySales
                    )
ORDER BY TRIM(Customer_ID);
```
