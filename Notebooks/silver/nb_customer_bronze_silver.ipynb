##### SILVER LAYER
nb_customer_bronze_silver

- Deleting Duplicated
- Column Schema Evolution
- Data Types Handling / Type Casting
- Null Records Handling
- Triming
- Case Conversion
- Maintaining Order of Data

```python
%sql
INSERT INTO intellibi_catalog.intellibi_silver.cleanedCustomer
SELECT DISTINCT
    TRIM(customer_id) AS customer_id,
    UPPER(TRIM(customer_name)) AS customer_name,
    UPPER(TRIM(email)) AS email,
    TRIM(phone) AS phone,
    UPPER(TRIM(address)) AS address,
    UPPER(TRIM(city)) AS city,
    UPPER(TRIM(state)) AS state,
    UPPER(TRIM(country)) AS country,
    TRIM(zip_code) AS zip_code,
    UPPER(TRIM(segment)) AS segment,
    CURRENT_TIMESTAMP() AS ingest_ts
FROM intellibi_catalog.intellibi_bronze.rawCustomer
WHERE ingest_ts > (
                    SELECT COALESCE(MAX(ingest_ts), TO_DATE('1900-01-01', 'yyyy-MM-dd')) 
                    FROM intellibi_catalog.intellibi_silver.cleanedCustomer
                    )
ORDER BY TRIM(customer_id);
```
