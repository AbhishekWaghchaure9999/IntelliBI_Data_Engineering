```python
%sql
-- inserting data from bronze to silver
INSERT INTO intellibi_catalog.intellibi_silver.cleanedpersons
SELECT DISTINCT
    UPPER(TRIM(person_name)) as person_name,
    CAST(TRIM(age) as int) AS age,
    UPPER(TRIM(gender)) AS gender,
    UPPER(TRIM(city)) AS city,
    current_timestamp AS ingest_ts
FROM intellibi_catalog.intellibi_bronze.rawPersons
WHERE ingest_ts > (
                    SELECT COALESCE(MAX(ingest_ts), TO_DATE('1900-01-01', 'yyyy-MM-dd')) 
                    FROM intellibi_catalog.intellibi_silver.cleanedPersons
                    )
ORDER BY TRIM(person_name);
```
