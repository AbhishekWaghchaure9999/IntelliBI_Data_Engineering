```python
%sql
MERGE INTO intellibi_catalog.intellibi_gold.refinedPersons As TGT
USING (
    SELECT * FROM intellibi_catalog.intellibi_silver.cleanedPersons
    where ingest_ts >
                ( 
                    SELECT COALESCE(MAX(last_update_ts), to_timestamp('1900-01-01'), 'YYYY-MM-DD')
                    FROM intellibi_catalog.intellibi_gold.refinedPersons
                )
) AS SRC
ON TGT.person_name = SRC.person_name
WHEN MATCHED AND 
(
    TGT.person_name <> SRC.person_name
    OR TGT.age <> SRC.age
    OR TGT.gender <> SRC.gender
    OR TGT.city <> SRC.city
)
THEN UPDATE SET
    TGT.person_name = SRC.person_name,
    TGT.age = SRC.age,
    TGT.gender = SRC.gender,
    TGT.city = SRC.city,
    TGT.intial_load_ts = current_timestamp,
    TGT.last_update_ts = current_timestamp
WHEN NOT MATCHED THEN INSERT (
    TGT.person_name,
    TGT.age,
    TGT.gender,
    TGT.city,
    TGT.intial_load_ts,
    TGT.last_update_ts
) VALUES (
    SRC.person_name,
    SRC.age,
    SRC.gender,
    SRC.city,
    current_timestamp,
    current_timestamp
)
```
